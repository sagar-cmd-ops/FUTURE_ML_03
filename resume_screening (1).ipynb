```python
import pandas as pd

data = {
    "resume": [
        "Python machine learning data analysis pandas numpy",
        "Java spring boot backend developer microservices",
        "Python deep learning tensorflow keras NLP",
        "SQL data analyst excel power bi tableau"
    ]
}

df = pd.DataFrame(data)
df
```

```python
job_description = "Python machine learning deep learning data science"
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    return text

df['clean_resume'] = df['resume'].apply(clean_text)
job_clean = clean_text(job_description)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

all_text = df['clean_resume'].tolist() + [job_clean]

tfidf_matrix = vectorizer.fit_transform(all_text)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

resume_vectors = tfidf_matrix[:-1]
job_vector = tfidf_matrix[-1]

scores = cosine_similarity(resume_vectors, job_vector)
```

```python
df['score'] = scores

df = df.sort_values(by='score', ascending=False)

df
```

```python
def skill_gap(resume, job):

    resume_words = set(resume.split())
    job_words = set(job.split())

    missing = job_words - resume_words

    return ", ".join(missing)

df['missing_skills'] = df['clean_resume'].apply(lambda x: skill_gap(x, job_clean))
df['score'] = (df['score'] * 100).round(2)
df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pickle

# ensure df is defined by running preceding cells
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['clean_resume'])

# save vectorizer
with open("vectorizer.pkl", "wb") as f:
    pickle.dump(vectorizer, f)

# save trained matrices or model
with open("tfidf_matrix.pkl", "wb") as f:
    pickle.dump(tfidf_matrix, f)
```
